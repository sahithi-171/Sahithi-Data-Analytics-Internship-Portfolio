# Key Performance Indicators (KPIs)

## 1. Total Titles
Formula: Count of all records

Purpose: Measures the size of Netflix's content library.

## 2. Total Movies
Formula: Count where Type = Movie

Purpose: Measures movie content availability.

## 3. Total TV Shows
Formula: Count where Type = TV Show

Purpose: Measures TV show availability.

## 4. Number of Countries
Formula: Count of unique countries

Purpose: Measures global content diversity.

## 5. Latest Release Year
Formula: Maximum release year

Purpose: Measures how current the catalog is.

```python
import pandas as pd

df = pd.read_csv("cleaned_netflix_data.csv")

df.head()
```

```python
print(df.columns)
```

```python
total_titles = len(df)
movies = (df["type"] == "Movie").sum()
tv_shows = (df["type"] == "TV Show").sum()
countries = df["country"].nunique()
latest_year = df["release_year"].max()

print("Total Titles:", total_titles)
print("Movies:", movies)
print("TV Shows:", tv_shows)
print("Countries:", countries)
print("Latest Release Year:", latest_year)
```
